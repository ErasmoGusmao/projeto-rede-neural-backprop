## Gradiente descendente - exemplo

$$
f(x) = x \cdot (x - 1) \cdot (x - 2) \cdot (x - 3)
$$

$$
f'(x) = 4x^3 - 18x^2 + 22x - 6
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x * (x - 1) * (x - 2) * (x - 3)

# Derivada de f(x)
def df(x):
    return 4*x**3 - 18*x**2 + 22*x - 6
```

### Cálculo do gradiente

```python
np.random.seed(42)

# Chute inicial
#x_hat = np.random.randn()*3
x_hat = 1.25

x = np.linspace(-0.1, 3.1, 100)

# Lista para guardar todos os valores de x_hat
x_hat_history = [x_hat]

for k in range(5):
    x_hat = x_hat - 0.1*df(x_hat)
    x_hat_history.append(x_hat)
```

### Gráfico de f(x) e do gradiente

```python
plt.plot(x, f(x), label='f(x)')

# Plotar o gradiente
for k, x_hat in enumerate(x_hat_history):
    plt.plot(x_hat, f(x_hat), 'ro')

    # Rótulo de cada ponto
    plt.annotate(
        f'{k}', 
        (x_hat, f(x_hat)), 
        xytext= (10,-3),
        textcoords='offset points')


plt.legend()
plt.show()
```

## Gradiente descendente – exemplo revisitado

```python
def df(x):
    
    # Caminho direto
    s0 = x
    s1 = x
    s2 = x
    s3 = x
    s4 = s1 - 1
    s5 = s2 - 2
    s6 = s3 - 3
    s7 = s0 * s4
    s8 = s5 * s6
    L = s7 * s8
    
    # Retornor (Gradiente)
    grad_L = 1
    grad_s7 = grad_L * s8
    grad_s8 = grad_L * s7
    grad_s6 = grad_s8 * s5
    grad_s5 = grad_s8 * s6
    grad_s4 = grad_s7 * s0
    grad_s3 = grad_s6
    grad_s2 = grad_s5
    grad_s1 = grad_s4
    grad_s0 = grad_s7 * s4
    grad_x = grad_s0 + grad_s1 + grad_s2 + grad_s3
    
    return grad_x
```
